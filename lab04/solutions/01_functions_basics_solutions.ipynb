## Our data

```python
# The genetic code: every three-letter codon -> its amino acid (one letter).
codon_table = {
    "TTT": "F", "TTC": "F", "TTA": "L", "TTG": "L",
    "CTT": "L", "CTC": "L", "CTA": "L", "CTG": "L",
    "ATT": "I", "ATC": "I", "ATA": "I", "ATG": "M",
    "GTT": "V", "GTC": "V", "GTA": "V", "GTG": "V",
    "TCT": "S", "TCC": "S", "TCA": "S", "TCG": "S",
    "CCT": "P", "CCC": "P", "CCA": "P", "CCG": "P",
    "ACT": "T", "ACC": "T", "ACA": "T", "ACG": "T",
    "GCT": "A", "GCC": "A", "GCA": "A", "GCG": "A",
    "TAT": "Y", "TAC": "Y", "TAA": "*", "TAG": "*",
    "CAT": "H", "CAC": "H", "CAA": "Q", "CAG": "Q",
    "AAT": "N", "AAC": "N", "AAA": "K", "AAG": "K",
    "GAT": "D", "GAC": "D", "GAA": "E", "GAG": "E",
    "TGT": "C", "TGC": "C", "TGA": "*", "TGG": "W",
    "CGT": "R", "CGC": "R", "CGA": "R", "CGG": "R",
    "AGT": "S", "AGC": "S", "AGA": "R", "AGG": "R",
    "GGT": "G", "GGC": "G", "GGA": "G", "GGG": "G",
}
stop_codons = {"TAA", "TAG", "TGA"}

# The same short gene, sampled from two related species.
species_a = "ATGAAAGGTTTTCTGGAATATGCAGGCTAA"   # the demos use this
species_b = "ATGAAAGGCTTCCTGGAGTATGCAAGCTAA"   # your exercises use this

print("species A:", species_a)
print("species B:", species_b)
```

## Solution — Skill 1, 🔬 write it: `at_content`

```python
def at_content(seq):
    at = 0
    for base in seq:
        if base in "AT":                # the only change from gc_content: the letters
            at = at + 1
    return round(100 * at / len(seq), 1)

at_b = at_content(species_b)
print(at_b)                             # 56.7
```

## Solution — Skill 1, 🐞 broken: `print` is not `return`

```python
def gc_content(seq):
    gc = 0
    for base in seq:
        if base in "GC":
            gc = gc + 1
    return round(100 * gc / len(seq), 1)   # return, not print: the caller gets the number back

if gc_content(species_b) > 40:
    print("species B is GC-rich")
```

## Solution — Skill 2, 🔬 write it: `base_percent`

```python
def base_percent(seq, base):
    total = 0
    for letter in seq:
        if letter == base:
            total = total + 1
    return round(total / len(seq) * 100, 1)   # only the name and this line changed

g_b = base_percent(species_b, "G")
print(g_b)                                    # 26.7
```

## Solution — Skill 3, 🐞 broken: a `return` in the wrong place

```python
def translate(seq):
    protein = ""
    for i in range(0, len(seq) - 2, 3):
        codon = seq[i:i + 3]
        if codon in stop_codons:
            break
        protein = protein + codon_table[codon]
    return protein                  # one indent to the left: AFTER the loop

print(translate(species_a))         # MKGFLEYAG
```

## Solution — Skill 3, 🔬 write it: `protein_differences`

```python
def protein_differences(seq_a, seq_b):
    protein_a = translate(seq_a)        # reuse translate - no new translation loop
    protein_b = translate(seq_b)
    differences = 0
    for i in range(len(protein_a)):
        if protein_a[i] != protein_b[i]:
            differences = differences + 1
    return differences

n_diff = protein_differences(species_a, species_b)
print(n_diff)                           # 1 - although the DNA differs in 4 places
```

## Solution — Skill 4, 🔬 write it: a default of your own

```python
def gc_content(seq, decimals=1):
    gc = 0
    for base in seq:
        if base in "GC":
            gc = gc + 1
    return round(100 * gc / len(seq), decimals)

print(gc_content(species_b))               # 43.3   <- the default
print(gc_content(species_b, 3))            # 43.333
print(gc_content(species_b, decimals=0))   # 43.0
```
