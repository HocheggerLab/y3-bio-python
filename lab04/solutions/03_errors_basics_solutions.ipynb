## Our data

```python
# The genetic code: every three-letter codon -> its amino acid (one letter).
codon_table = {
    "TTT": "F", "TTC": "F", "TTA": "L", "TTG": "L",
    "CTT": "L", "CTC": "L", "CTA": "L", "CTG": "L",
    "ATT": "I", "ATC": "I", "ATA": "I", "ATG": "M",
    "GTT": "V", "GTC": "V", "GTA": "V", "GTG": "V",
    "TCT": "S", "TCC": "S", "TCA": "S", "TCG": "S",
    "CCT": "P", "CCC": "P", "CCA": "P", "CCG": "P",
    "ACT": "T", "ACC": "T", "ACA": "T", "ACG": "T",
    "GCT": "A", "GCC": "A", "GCA": "A", "GCG": "A",
    "TAT": "Y", "TAC": "Y", "TAA": "*", "TAG": "*",
    "CAT": "H", "CAC": "H", "CAA": "Q", "CAG": "Q",
    "AAT": "N", "AAC": "N", "AAA": "K", "AAG": "K",
    "GAT": "D", "GAC": "D", "GAA": "E", "GAG": "E",
    "TGT": "C", "TGC": "C", "TGA": "*", "TGG": "W",
    "CGT": "R", "CGC": "R", "CGA": "R", "CGG": "R",
    "AGT": "S", "AGC": "S", "AGA": "R", "AGG": "R",
    "GGT": "G", "GGC": "G", "GGA": "G", "GGG": "G",
}
stop_codons = {"TAA", "TAG", "TGA"}

# The same short gene, sampled from two related species.
species_a = "ATGAAAGGTTTTCTGGAATATGCAGGCTAA"   # the demos use this
species_b = "ATGAAAGGCTTCCTGGAGTATGCAAGCTAA"   # your exercises use this

print("species A:", species_a)
print("species B:", species_b)


# our translate() from the functions notebooks
def translate(seq, frame=0):
    protein = ""
    for i in range(frame, len(seq) - 2, 3):
        codon = seq[i:i + 3]
        if codon in stop_codons:
            break
        protein = protein + codon_table[codon]
    return protein

# ...and some MESSY real-world sequences to break it on
messy = "ATGAANGGTTTT"        # an 'N' - the sequencer wasn't sure of this base
lower = "atgaaaggttttctg"     # lower-case letters
```

## Solution — Skill 1: read a traceback

```python
# translate(lower) crashes. The last line of the traceback reads:
#
#     KeyError: 'atg'
#
# The error type is KeyError, and the "codon" that upset it is 'atg'.
# Why: the keys in codon_table are UPPER-case ("ATG"), and to a dictionary
# 'atg' and 'ATG' are two different keys - so 'atg' isn't there.

print("atg" in codon_table, "ATG" in codon_table)   # False True
```

## Solution — Skill 2: `try` / `except`

```python
try:
    protein = translate(lower)
    print("protein:", protein)
except KeyError:
    print("couldn't translate that one - unknown codon")

print("done")                           # still runs: the program survived
```

## Solution — Skill 3: catch the right error

```python
# safe_translate from the demo
def safe_translate(seq):
    try:
        return translate(seq)
    except KeyError:
        return "?"

for seq in [species_a, species_b, messy, lower]:
    print(seq, "->", safe_translate(seq))

# species_a and species_b give a protein; messy (an N) and lower (lower-case) give "?".
# The downside: "?" hides WHY it failed. lower-case DNA is perfectly good data that
# .upper() would fix - but it gets thrown away with the truly broken sequence.
```

## Solution — Skill 4: `raise` your own

```python
def check_length(seq):
    if len(seq) % 3 != 0:
        raise ValueError(f"length {len(seq)} isn't a whole number of codons")
    return seq

print(check_length(species_a))          # length 30: fine, comes back unchanged

try:
    check_length("ATGAA")               # length 5: raises
except ValueError as e:
    print("ValueError:", e)             # (the try/except is only so this notebook runs on)
```
