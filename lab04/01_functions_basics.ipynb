## Our data — one gene, two species

Run this first. Every exercise below uses these.

- `ancestor` — the gene in **species A** (the demos use this)
- `species_b` — the same gene in **species B** (your DIY challenges use this)
- `codon_table` — the genetic code: codon → amino-acid letter

````python
# The genetic code: every three-letter codon -> its amino acid (one letter).
codon_table = {
    "TTT": "F", "TTC": "F", "TTA": "L", "TTG": "L",
    "CTT": "L", "CTC": "L", "CTA": "L", "CTG": "L",
    "ATT": "I", "ATC": "I", "ATA": "I", "ATG": "M",
    "GTT": "V", "GTC": "V", "GTA": "V", "GTG": "V",
    "TCT": "S", "TCC": "S", "TCA": "S", "TCG": "S",
    "CCT": "P", "CCC": "P", "CCA": "P", "CCG": "P",
    "ACT": "T", "ACC": "T", "ACA": "T", "ACG": "T",
    "GCT": "A", "GCC": "A", "GCA": "A", "GCG": "A",
    "TAT": "Y", "TAC": "Y", "TAA": "*", "TAG": "*",
    "CAT": "H", "CAC": "H", "CAA": "Q", "CAG": "Q",
    "AAT": "N", "AAC": "N", "AAA": "K", "AAG": "K",
    "GAT": "D", "GAC": "D", "GAA": "E", "GAG": "E",
    "TGT": "C", "TGC": "C", "TGA": "*", "TGG": "W",
    "CGT": "R", "CGC": "R", "CGA": "R", "CGG": "R",
    "AGT": "S", "AGC": "S", "AGA": "R", "AGG": "R",
    "GGT": "G", "GGC": "G", "GGA": "G", "GGG": "G",
}
stop_codons = {"TAA", "TAG", "TGA"}

# The same short gene, sampled from two related species.
ancestor  = "ATGAAAGGTTTTCTGGAATATGCAGGCTAA"   # species A - the demos use this
species_b = "ATGAAAGGCTTCCTGGAGTATGCAAGCTAA"   # species B - your DIY challenges use this

print("species A:", ancestor)
print("species B:", species_b)
````

## Skill 1 — Define a function & `return` an answer

A **function** is a named recipe. You **define** it once, then **call** it as often as you like:

```python
def gc_content(seq):
    ...
    return answer
```

- `def` starts the definition; `gc_content` is the name; `seq` is the **input** (a *parameter*)
- the indented lines are the **body** — they only run when you *call* it
- `return` hands an answer **back** to whoever called it

The **GC content** — the fraction of a sequence that is G or C — is a classic first measurement in evolution: it drifts between species.

````python
# DEMO - measure the GC content (% of bases that are G or C)
def gc_content(seq):
    gc = 0
    for base in seq:
        if base in "GC":
            gc = gc + 1
    return round(100 * gc / len(seq), 1)

# define once, call as often as we like
print("species A GC%:", gc_content(ancestor))
````

## Skill 2 — Parameters: one function, many questions

A function gets more useful when it takes **more than one** input. Extra parameters let the *same* recipe answer *many* questions:

```python
def count_base(seq, base):
    ...
```

Now `count_base(seq, "A")` and `count_base(seq, "G")` both work — you choose what to count when you **call** it.

````python
# DEMO - count how many times one base appears
def count_base(seq, base):
    total = 0
    for letter in seq:
        if letter == base:
            total = total + 1
    return total

print("A in species A:", count_base(ancestor, "A"))
print("T in species A:", count_base(ancestor, "T"))
````

## Skill 3 — Build `translate()` — your biggest tool yet

Time to package the whole DNA→protein recipe behind one name. You already have every piece from Lab 3:

- **loop** over the sequence in steps of **3** — one codon at a time
- **slice** out each codon: `seq[i:i+3]`
- **look it up** in `codon_table`
- **stop** at a stop codon, and `return` the protein

```python
for i in range(0, len(seq) - 2, 3):
    codon = seq[i:i + 3]
```

`range(0, len(seq) - 2, 3)` gives you `0, 3, 6, …` — the start of each codon.

````python
# DEMO - translate DNA into protein
def translate(seq):
    protein = ""
    for i in range(0, len(seq) - 2, 3):
        codon = seq[i:i + 3]
        if codon in stop_codons:
            break                       # stop codon - end of the protein
        protein = protein + codon_table[codon]
    return protein

print("species A DNA    :", ancestor)
print("species A protein:", translate(ancestor))
````

## Skill 4 — Default arguments: sensible defaults you can override

A parameter can have a **default value**, used when the caller doesn't supply one:

```python
def translate(seq, frame=0):
    ...
```

A **reading frame** is where you start counting codons: frame `0`, `1`, or `2`. Shift the frame and you read a completely different protein — biology cares a lot about which frame a gene uses. With a default of `0`, old calls like `translate(seq)` still work unchanged.

````python
# DEMO - translation with a choosable reading frame (default 0)
def translate(seq, frame=0):
    protein = ""
    for i in range(frame, len(seq) - 2, 3):
        codon = seq[i:i + 3]
        if codon in stop_codons:
            break
        protein = protein + codon_table[codon]
    return protein

print("frame 0 (default):", translate(ancestor))       # same as before
print("frame 1          :", translate(ancestor, 1))     # start one base later
print("frame 2          :", translate(ancestor, frame=2))
````
