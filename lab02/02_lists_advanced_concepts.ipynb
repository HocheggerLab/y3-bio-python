## 1 — Nested lists

A list can hold **other lists**. That's how you store a table: here, three quadrats, each with the counts of oak, birch and hazel.

- `quadrats[0]` → the whole **first quadrat** (a list)
- `quadrats[0][2]` → **row 0, column 2** → hazel in the first quadrat
- `sum(quadrats[1])` → total individuals in the second quadrat

```python
# DEMO — a grid of counts:      oak, birch, hazel
quadrats = [
    [34, 12, 5],   # quadrat 0
    [40,  9, 7],   # quadrat 1
    [28, 15, 4],   # quadrat 2
]

print("first quadrat:", quadrats[0])
print("hazel in quadrat 0:", quadrats[0][2])
print("total in quadrat 1:", sum(quadrats[1]))
print("number of quadrats:", len(quadrats))
```

## 2 — The copy-vs-alias gotcha

This one bites everyone once. When you write `b = a`, you do **not** get a second list — `a` and `b` are two names for the **same** list. Change one, you change both.

To get a real, independent copy, use `.copy()` (or a full slice `a[:]`).

```python
# DEMO — the trap
original = [34, 12, 5]
backup = original          # NOT a copy — same list, two names

backup[0] = 999
print("backup:  ", backup)
print("original:", original)   # 😱 it changed too!
```

```python
# DEMO — the fix
original = [34, 12, 5]
backup = original.copy()   # a real, independent copy

backup[0] = 999
print("backup:  ", backup)
print("original:", original)   # 😊 safe
```

## 3 — What else? Pairing, ranking, and a peek ahead

Three genuinely useful tools once your data grows:

- **`zip()`** — stitch two lists together, item by item. Perfect for `species` + `counts`.
- **`sorted(…, key=…)`** — sort by something you choose, e.g. rank species by abundance.
- **List comprehension** — a one-line way to build a new list. You'll fully meet the `for` next session; here's a taste.

```python
# DEMO — pair species with their counts
species = ["oak", "birch", "hazel", "rowan", "holly", "bramble"]
counts  = [34, 12, 5, 8, 3, 27]

pairs = list(zip(species, counts))
print("paired:", pairs)

# rank species by abundance, most first
ranked = sorted(pairs, key=lambda pair: pair[1], reverse=True)
print("most abundant first:", ranked)
print("the winner:", ranked[0])
```

```python
# DEMO — a peek at a list comprehension (next session you'll read the `for` inside)
doubled = [c * 2 for c in counts]
print("counts doubled:", doubled)
```
