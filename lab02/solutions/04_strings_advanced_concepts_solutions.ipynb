## Solution — 1: slicing tricks

```python
scientific = "Quercus robur"

print(scientific[::-1])        # rubor sucreuQ
print("SITE04"[::2])           # ST0

dna = "AATTCCGG"
print(dna[::-1])               # GGCCTTAA
```

## Solution — 2: the immutability gotcha

```python
messy = "  Birch "
messy.lower()                  # returns a new string… which we throw away
print(f"[{messy}]")            # [  Birch ] — unchanged

messy = messy.strip().lower()  # catch the result by reassigning
print(f"[{messy}]")            # [birch]

capitalised = messy[0].upper() + messy[1:]   # build a new string
print(capitalised)             # Birch
```

## Solution — 3: tables & chaining

```python
print(f"{'species':<10}{'count':>6}")
print(f"{'rowan':<10}{8:>6}")
print(f"{'bramble':<10}{27:>6}")

print("  hazel coppice  ".strip().title())    # Hazel Coppice

# Stretch — maxsplit=1 splits only at the first comma
print("oak,34,north-slope,2023".split(",", maxsplit=1))
# ['oak', '34,north-slope,2023']
```
