## Our data

```python
species = ["oak", "birch", "hazel", "rowan", "holly", "bramble"]
counts  = [34, 12, 5, 8, 3, 27]

years = [2018, 2019, 2020, 2021, 2022, 2023, 2024]
pop   = [120, 135, 150, 145, 160, 158, 170]
```

## Solution — Skill 1: reach in

```python
print("oak count:", counts[0])
print("last three species:", species[-3:])
print("every 2nd count:", counts[::2])
print("wolf recorded?", "wolf" in species)      # False
```

## Solution — Skill 2: change values

```python
counts[4] = 9          # holly was misread: 3 → 9
print(counts)

del counts[-1]         # bramble was logged twice — drop the last count
print(counts)
```

## Solution — Skill 3: grow and combine

```python
new_species = ["willow", "elder"]

species.append("willow")                 # add one item, in place
full_survey = species + new_species      # a brand-new combined list
print(full_survey)
print("how many:", len(full_survey))

# willow now appears twice — once from .append(), once from new_species.
# + doesn't check for duplicates; it just glues the lists together.
```

## Solution — Skill 4: summarise

```python
total = sum(pop)
mean = total / len(pop)

print("total:", total)
print(f"mean population: {mean:.1f}")
print("highest:", max(pop))
print("lowest: ", min(pop))
```

## Solution — Skill 5: sort by value

```python
print("low → high:", sorted(pop))
print("high → low:", sorted(pop, reverse=True))
print("unchanged: ", pop)          # sorted() returns a new list
```

## 🔭 Explore further

```python
# Strings sort alphabetically (A–Z)
print(sorted(species))

# Mixed types can't be compared, so sorting fails:
# sorted([3, "oak", 1.5])   # ← TypeError: '<' not supported between 'str' and 'int'
# ints and floats mix fine, though — they're both numbers:
print(sorted([3, 1.5, 2]))
```
