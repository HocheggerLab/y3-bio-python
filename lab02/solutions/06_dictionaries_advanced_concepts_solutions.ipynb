## Our data

```python
abundance = {"oak": 34, "birch": 12, "hazel": 5, "rowan": 8, "holly": 3, "bramble": 27}

trees = {
    "oak":   {"scientific": "Quercus robur",  "native": True,  "height_m": 28},
    "birch": {"scientific": "Betula pendula", "native": True,  "height_m": 20},
}
sites = {
    "north-slope": ["oak", "birch"],
    "east-ridge":  ["hazel", "rowan", "holly"],
}
```

## Solution — 1: nested dicts & dicts of lists

```python
print(trees["birch"]["scientific"])     # Betula pendula
print(trees["oak"]["native"])           # True
print(len(sites["north-slope"]))        # 2
```

## Solution — 2: the keys gotcha

```python
print({"hazel": 5, "hazel": 9})        # {'hazel': 9} — the last one wins

# Why? A dict looks values up *by key*. If two entries shared a key,
# abundance["hazel"] couldn't know which value to give back — so a key
# can only ever point at one value, and re-using it overwrites.

# bad = {["north", "slope"]: 5}         # TypeError: unhashable type: 'list'
good = {("north", "slope"): 5}          # a tuple can't change, so it's a valid key
print(good)
```

## Solution — 3: build, rank & comprehension

```python
birds = ["robin", "wren", "tit"]
seen = [4, 2, 6]
print(dict(zip(birds, seen)))

ranked = sorted(abundance.items(), key=lambda kv: kv[1])   # reverse=False is the default
print(ranked)

# Stretch
doubled = {name: n * 2 for name, n in abundance.items()}
print(doubled)
```

## 🔭 Explore — the codon table

```python
codon_table = {"ATG": "Met", "TTT": "Phe", "GGG": "Gly", "TAA": "Stop"}

print("TTT →", codon_table["TTT"])
print("TAA →", codon_table["TAA"])
print("CGA →", codon_table.get("CGA", "?"))   # not in our mini table
```
