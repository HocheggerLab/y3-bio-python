## Our data

```python
abundance = {"oak": 34, "birch": 12, "hazel": 5, "rowan": 8, "holly": 3, "bramble": 27}
```

## Solution — Skill 1: look up by key

```python
hazel = abundance["hazel"]
holly = abundance["holly"]
print(f"We counted {hazel} hazel and {holly} holly.")
```

## Solution — Skill 2: add, update & remove

```python
abundance["willow"] = 6       # new key → added
abundance["oak"] = 40         # existing key → overwritten
del abundance["bramble"]
print(abundance)
```

## Solution — Skill 3: safe lookup

```python
print("badger:", abundance.get("badger", 0))   # 0 — no KeyError

print("hazel" in abundance)                    # True

if "hazel" in abundance:
    print("Yes — hazel was recorded")
else:
    print("No — hazel was not recorded")
```

## Solution — Skill 4: look through it

```python
print(list(abundance.keys()))
print("total:", sum(abundance.values()))
print("species:", len(abundance))

# Stretch
print(list(abundance.items()))
```
