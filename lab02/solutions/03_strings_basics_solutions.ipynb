## Our data

```python
sample_code = "SITE04-OAK-2023"
record      = "oak,34,north-slope"
scientific  = "Quercus robur"
messy_names = ["  oak ", "BIRCH", "Hazel  ", "rowan"]
```

## Solution — Skill 1: index & slice

```python
my_code = "SITE12-BIRCH-2024"

print(my_code[0])        # S
print(my_code[:6])       # SITE12
print(my_code[-4:])      # 2024
print(len(my_code))      # 17
```

## Solution — Skill 2: clean up

```python
print(messy_names[2].strip())               # Hazel
print(messy_names[1].lower())               # birch
print("  quercus robur ".strip().title())   # Quercus Robur
```

## Solution — Skill 3: search & test

```python
my_code = "SITE12-BIRCH-2024"

print("BIRCH" in my_code)          # True
print(my_code.endswith("2024"))    # True
print(my_code.count("-"))          # 2
print(my_code.find("1"))           # 4 — counting from 0: S0 I1 T2 E3 '1'4
```

## Solution — Skill 4: build & combine

```python
species = "birch"
count = 12
place = "east-ridge"

print(f"{species.title()} — {count} counted at {place}")
print(place.replace("east", "west"))      # west-ridge
```

## Solution — Skill 5: split & join

```python
genus, species = scientific.split(" ")    # two pieces → two names
print(genus)                               # Quercus

sites = ["north-slope", "east-ridge", "west-fen"]
print(" → ".join(sites))
```
