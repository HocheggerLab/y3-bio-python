## Solution — 1: nested lists

```python
quadrats = [
    [34, 12, 5],   # quadrat 0
    [40,  9, 7],   # quadrat 1
    [28, 15, 4],   # quadrat 2
]

print("birch in quadrat 2:", quadrats[2][1])        # 15
print("total in last quadrat:", sum(quadrats[-1]))  # 47

# Stretch — total oak across all quadrats (column 0 of every row)
print("total oak:", quadrats[0][0] + quadrats[1][0] + quadrats[2][0])   # 102
```

## Solution — 2: copy vs alias

```python
counts = [34, 12, 5, 8, 3, 27]

# The trap — snapshot is just a second name for the same list
snapshot = counts
counts.append(50)
print("snapshot (alias):", snapshot)   # 50 shows up here too!

# Done properly — copy *before* changing
counts = [34, 12, 5, 8, 3, 27]
snapshot = counts.copy()               # or counts[:]
counts.append(50)
print("counts:  ", counts)
print("snapshot:", snapshot)           # still the original six
```

## Solution — 3: pairing and ranking

```python
years = [2018, 2019, 2020, 2021, 2022, 2023, 2024]
pop   = [120, 135, 150, 145, 160, 158, 170]

pairs = list(zip(years, pop))
ranked = sorted(pairs, key=lambda pair: pair[1], reverse=True)
print("ranked:", ranked)
print("biggest population:", ranked[0])         # (2024, 170)
print("year:", ranked[0][0])

# Stretch — counts as percentages of the total
counts = [34, 12, 5, 8, 3, 27]
total = sum(counts)
percent = [c / total * 100 for c in counts]
print([round(p, 1) for p in percent])
```
