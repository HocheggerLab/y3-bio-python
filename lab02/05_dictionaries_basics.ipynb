## Our data — species paired with their counts

Run this first. A dictionary lives in `{curly braces}`, as `key: value` pairs.

```python
abundance = {"oak": 34, "birch": 12, "hazel": 5, "rowan": 8, "holly": 3, "bramble": 27}
print(abundance)
```

## Skill 1 — Make a dict & look up by key

In a list you reach in by **position**. In a dictionary you reach in by **key** — here, the species name:

- `abundance["oak"]` → `34`
- keys are **unique** — each species appears once
- `list(abundance)` → all the keys

```python
# DEMO — look things up by name, not by position
print("oaks:    ", abundance["oak"])
print("brambles:", abundance["bramble"])
print("all species:", list(abundance))   # the keys
```

## Skill 2 — Add, update & remove entries

Dictionaries are **mutable**:

- `abundance["gorse"] = 15` → assign a **new** key to **add**
- `abundance["holly"] = 9` → assign an **existing** key to **overwrite**
- `del abundance["rowan"]` → remove an entry

```python
# DEMO — the survey gets corrected
abundance["gorse"] = 15        # a new species turned up
abundance["holly"] = 9         # holly was miscounted (was 3)
del abundance["rowan"]         # rowan was a misidentification

print(abundance)
```

## Skill 3 — Safe lookup

Look up a key that isn't there and Python **crashes** with a `KeyError`. Two safe ways round it:

- `"wolf" in abundance` → `True` / `False` (check first)
- `abundance.get("wolf", 0)` → the value, or a **default** (`0`) if the key is missing

```python
# DEMO — a species you never recorded
# print(abundance["wolf"])                    # ← this would crash: KeyError

print("wolf recorded?", "wolf" in abundance)         # False
print("wolf count:   ", abundance.get("wolf", 0))    # 0 — a safe default
print("oak count:    ", abundance.get("oak", 0))     # 34
```

## Skill 4 — Look through it

Get at everything at once:

- `.keys()` → the species &nbsp;•&nbsp; `.values()` → the counts &nbsp;•&nbsp; `.items()` → `(species, count)` pairs
- `sum(abundance.values())` → total individuals
- `len(abundance)` → how many species

```python
# DEMO — summarise the whole survey
print("species:", list(abundance.keys()))
print("counts: ", list(abundance.values()))
print("total individuals:", sum(abundance.values()))
print("number of species:", len(abundance))
```
