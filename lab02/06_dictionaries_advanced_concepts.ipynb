```python
abundance = {"oak": 34, "birch": 12, "hazel": 5, "rowan": 8, "holly": 3, "bramble": 27}

trees = {
    "oak":   {"scientific": "Quercus robur",  "native": True,  "height_m": 28},
    "birch": {"scientific": "Betula pendula", "native": True,  "height_m": 20},
}
sites = {
    "north-slope": ["oak", "birch"],
    "east-ridge":  ["hazel", "rowan", "holly"],
}
```

## 1 — Nested dicts & dicts of lists

A value can be **anything** — including another dict or a list. That's how you store structured records (a dict per species) or grouped data (a list per site). Reach in by **chaining** keys:

- `trees["oak"]["height_m"]` → `28`
- `sites["east-ridge"]` → a list; `len(...)` counts it

```python
# DEMO — reach into structured data
print("oak's Latin name:", trees["oak"]["scientific"])
print("oak's height:    ", trees["oak"]["height_m"], "m")

print("species on east-ridge:", sites["east-ridge"])
print("how many there:       ", len(sites["east-ridge"]))
```

## 2 — The keys gotcha

Keys must be **unique** and **immutable**.

- Write the same key twice and the **last one silently wins** — you can never have two entries under one key. (That's exactly *why* assigning to an existing key updates rather than adds.)
- A key must be immutable: a string, number or tuple is fine — a **list cannot be a key** (`TypeError: unhashable type: 'list'`).

```python
# DEMO — a duplicate key silently overwrites
counts = {"oak": 1, "oak": 2}
print(counts)                       # {'oak': 2} — only one 'oak', last wins

# DEMO — a list can't be a key
# bad = {["north", "slope"]: 5}     # ← uncomment to see: TypeError: unhashable type: 'list'
good = {("north", "slope"): 5}      # a tuple is fine — it's immutable
print(good)
```

## 3 — What else? Build, rank, and a comprehension teaser

- **`dict(zip(keys, values))`** — build a dict straight from two parallel lists (the callback to your lists & strings work).
- **`sorted(d.items(), key=…, reverse=True)`** — rank a dict by its **values**.
- **Dict comprehension** — build a whole new dict in one line (a taste of next session's `for`).

```python
# DEMO — build a dict from two parallel lists
species = ["oak", "birch", "hazel"]
nums    = [34, 12, 5]
built = dict(zip(species, nums))
print("built:", built)

# DEMO — rank species by abundance, most first
ranked = sorted(abundance.items(), key=lambda kv: kv[1], reverse=True)
print("ranked:", ranked)
print("most abundant:", ranked[0])

# DEMO — a dict comprehension: each count as a % of the total
total = sum(abundance.values())
percent = {name: round(n / total * 100, 1) for name, n in abundance.items()}
print("as % of total:", percent)
```

## 🔭 Explore further — the codon table

The dictionary you've been building is *exactly* how bioinformatics stores the **genetic code**: a `codon → amino acid` lookup. And `.get(codon, "?")` is what keeps a real translator from crashing on a bad codon.

```python
# 🔭 EXPLORE — a tiny slice of the codon table
codon_table = {
    "ATG": "Met",    # start
    "TTT": "Phe",
    "GGG": "Gly",
    "TAA": "Stop",
}
print("ATG →", codon_table["ATG"])
print("GGG →", codon_table["GGG"])
print("NNN →", codon_table.get("NNN", "?"))   # unknown codon — stays safe
```
