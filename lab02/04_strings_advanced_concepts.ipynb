## 1 — Slicing tricks

Strings take the full `text[start:stop:step]` slice — same as lists:

- `text[::2]` → every 2nd character
- `text[::-1]` → the whole thing **reversed**

Reversing a sequence is a genuine bioinformatics move (the first step toward a reverse-complement).

```python
# DEMO — steps and reverse
code = "SITE04-OAK-2023"
print("every 2nd char:", code[::2])
print("reversed:      ", code[::-1])

dna = "ATGCGT"
print("reversed DNA:  ", dna[::-1])
```

## 2 — The immutability gotcha

Strings are **immutable** — you cannot change them in place. This trips up everyone once:

- `name[0] = "X"` → **`TypeError`**
- A method like `name.strip()` does **not** change `name` — it **returns a new string** you must catch by reassigning.

```python
# DEMO — the trap: methods don't change the original
name = "  oak "
name.strip()            # looks like it cleans it...
print(f"[{name}]")      # ...but name is unchanged!  ->  [  oak ]
```

```python
# DEMO — the fix: reassign the result
name = "  oak "
name = name.strip()     # catch the new string
print(f"[{name}]")      # ->  [oak]
```

```python
# DEMO — you can't edit one character in place
name = "oak"
# name[0] = "O"         # ← uncomment to see: TypeError
name = "O" + name[1:]   # instead, build a new string
print(name)             # Oak
```

## 3 — What else? Tidy tables, split/join power, chaining

- **f-string formatting** lines columns up: `{value:<10}` left-aligns in 10 spaces, `{value:>6}` right-aligns in 6, `{value:.1f}` rounds.
- **`.split(sep, maxsplit=…)`** limits how many splits you make.
- **Method chaining** — each string method returns a string, so you can stack them.

```python
# DEMO — a tidy aligned table with f-strings
species = ["oak", "birch", "hazel"]
counts  = [34, 12, 5]

print(f"{'species':<10}{'count':>6}")
print(f"{species[0]:<10}{counts[0]:>6}")
print(f"{species[1]:<10}{counts[1]:>6}")
print(f"{species[2]:<10}{counts[2]:>6}")
```

```python
# DEMO — chain methods in one go
raw = "  OAK-woodland  "
clean = raw.strip().lower().replace("-", " ")
print(f"[{clean}]")     # [oak woodland]
```

## 🔭 Explore further — from field notes back to DNA

The exact slicing you just used works on **sequence data** too. A gene is read in **codons** — three letters at a time — which is just `seq[i:i+3]` for `i = 0, 3, 6, …`

```python
# 🔭 EXPLORE — slice a DNA sequence into codons
seq = "ATGCGTAAT"
print("codon 1:", seq[0:3])   # ATG
print("codon 2:", seq[3:6])   # CGT
print("codon 3:", seq[6:9])   # AAT
```
