## Skill 1 — A string is a sequence

A string is just a **sequence of characters** — like a list, but of letters. So everything you learned about indexing and slicing lists works here too, counting **from 0**:

- `sample_code[0]` → the first character
- `sample_code[:6]` → a slice: the first six characters
- `sample_code[-4:]` → the last four (the year)
- `len(sample_code)` → how many characters

```python
# DEMO — pick apart a sample code: SITE04-OAK-2023
print("first character:", sample_code[0])
print("site part:      ", sample_code[:6])    # SITE04
print("the year:       ", sample_code[-4:])   # 2023
print("code length:    ", len(sample_code))
```

## Skill 2 — Change case & clean up

Field data is messy — stray spaces and random capitals. These methods hand you a **tidied-up copy**:

- `.strip()` → remove spaces from the ends
- `.upper()` / `.lower()` → force one case
- `.title()` → Capitalise Each Word

```python
# DEMO — tidy one messy entry: "  oak "
raw = messy_names[0]
print(f"raw:      [{raw}]")
print(f"stripped: [{raw.strip()}]")
print("upper:   ", raw.strip().upper())
print("title:   ", raw.strip().title())
```

## Skill 3 — Search & test

Ask questions of a string:

- `"OAK" in sample_code` → `True` / `False`
- `sample_code.find("-")` → position of the first match (`-1` if not found)
- `sample_code.startswith("SITE")` / `.endswith("2023")` → `True` / `False`
- `sample_code.count("-")` → how many times something appears

```python
# DEMO — interrogate SITE04-OAK-2023
print("has OAK?        ", "OAK" in sample_code)
print("position of '-':", sample_code.find("-"))     # 6
print("from 2023?      ", sample_code.endswith("2023"))
print("how many '-':   ", sample_code.count("-"))    # 2
```

## Skill 4 — Build & combine

Stitch strings together:

- `+` glues strings, but an **f-string** is cleaner: `f"{species}: {count}"`
- `.replace(old, new)` swaps text for a corrected copy

> ⚠️ Strings are **immutable** — `.replace()` returns a *new* string; the original is left alone.

```python
# DEMO — build a label, and fix a location
species = "oak"
count = 34
label = f"{species.title()}: {count} individuals"
print(label)

fixed = record.replace("north", "south")
print("before:", record)    # unchanged — strings are immutable
print("after: ", fixed)
```

## Skill 5 — Split & join

The bridge between **text** and **lists**:

- `record.split(",")` → breaks a string into a **list** of pieces
- `" ".join(list)` → glues a list of strings back into **one string**

```python
# DEMO — a field record is really three fields
fields = record.split(",")
print("as a list:", fields)                  # ['oak', '34', 'north-slope']
print("species:", fields[0], " count:", fields[1])

# join a list of names into a sentence
names = ["oak", "birch", "hazel"]
sentence = ", ".join(names)
print("joined:", sentence)                   # oak, birch, hazel
```
