## Our data — a woodland survey

Run this first. Every exercise below uses these lists.

- `species` — the plants we recorded, in the order we found them
- `counts` — how many of each we counted in one quadrat (same order as `species`)
- `years` / `pop` — a separate record: a deer population counted once a year

```python
species = ["oak", "birch", "hazel", "rowan", "holly", "bramble"]
counts  = [34, 12, 5, 8, 3, 27]

years = [2018, 2019, 2020, 2021, 2022, 2023, 2024]
pop   = [120, 135, 150, 145, 160, 158, 170]   # deer counted each year

print(species)
print(counts)
```

## Skill 1 — Make a list and reach in

A **list** holds many values in order, inside square brackets `[ ]`. You reach in by **position** (the *index*), counting **from 0**:

- `species[0]` → the **first** species &nbsp;•&nbsp; `species[-1]` → the **last**
- `species[1:3]` → a **slice**: from index 1 up to *but not including* 3
- `pop[::2]` → a **step**: every 2nd value (great for alternate years)
- `len(species)` → how many items &nbsp;•&nbsp; `"hazel" in species` → `True` / `False`

```python
# DEMO — reaching into the survey
print("first species:", species[0])
print("last species: ", species[-1])
print("middle slice: ", species[1:3])      # birch, hazel
print("every 2nd year pop:", pop[::2])     # 2018, 2020, 2022, 2024
print("reversed counts:", counts[::-1])
print("how many species:", len(species))
print("did we see hazel?", "hazel" in species)
```

## Skill 2 — Change values in a list

Lists are **mutable** — you can change them after making them.

- `species[3] = "ash"` → overwrite the value at position 3
- `del counts[0]` → remove the item at a position

> ⚠️ Changing by position changes the list *in place* — the old value is gone.

```python
# DEMO — you realise the 'rowan' was actually an ash
print("before:", species)
species[3] = "ash"
print("after: ", species)
```

## Skill 3 — Grow and combine lists

Surveys grow. You can add to a list as you go:

- `species.append("gorse")` → add **one** item to the end
- `species.insert(0, "willow")` → add at a **position**
- `list_a + list_b` → **join** two lists into a new one
- `species.extend(more)` → add every item from another list onto the end

```python
# DEMO — a second quadrat turned up two new species
found = ["gorse", "heather"]

survey = species + found          # a brand-new combined list
print("combined:", survey)
print("how many now:", len(survey))

species.append("gorse")           # add just one, in place
print("after append:", species)
```

## Skill 4 — Summarise the numbers

When a list holds numbers, Python can summarise the whole thing at once — **no loop needed** (that power comes next session):

- `sum(counts)` → add them all up &nbsp;•&nbsp; `len(counts)` → how many
- `sum(counts) / len(counts)` → the **mean**
- `min(counts)` / `max(counts)` → smallest / largest
- `sorted(counts)` → a new list, smallest → largest

```python
# DEMO — describe the quadrat
total = sum(counts)
mean = total / len(counts)

print("total individuals:", total)
print(f"mean abundance:   {mean:.1f}")
print("least abundant count:", min(counts))
print("most abundant count: ", max(counts))
print("sorted low→high:", sorted(counts))
```
