## Skill 1 — `while` loops

A `for` loop needs to know the items in advance. But sometimes you **don't know how many steps** something will take — you only know the **condition** to stop.

Classic neuroscience example: **integrate-and-fire**. A neuron adds up incoming current; its membrane voltage climbs; the instant it crosses **threshold**, it fires. How many steps? Depends on the current — so we loop *while* it's still below threshold.

```python
while <condition is True>:
    <do one step>
```

````python
# DEMO — integrate-and-fire: how many steps until this neuron fires?
voltage   = 0.0    # start at rest
threshold = 5.0    # fires when voltage reaches this
current   = 1.2    # input added each step
steps     = 0

while voltage < threshold:
    voltage += current      # integrate one step of input
    steps   += 1            # count the step

print(f"fired after {steps} steps (V = {voltage:.1f})")
````

## Skill 2 — `break`: stop early

Once you've **found what you're looking for**, there's no point continuing. `break` jumps straight out of the loop.

Here: find the **first burst** (a bin with more than 8 spikes) and stop scanning.

````python
# DEMO — find the first burst, then stop
for i, count in enumerate(spikes):
    if count > 8:
        print("first burst at bin", i, "with", count, "spikes")
        break            # found it — don't look at the rest
````

> ⚠️ **Infinite loops.** A `while` whose condition never becomes `False` runs forever. `break` is one safe escape hatch:
> ```python
> while True:          # looks infinite…
>     ...
>     if done:
>         break        # …but this gets us out
> ```
> If you *do* get stuck in one, stop it with the ⏹ **Interrupt** button (or *Kernel → Interrupt*).

## Skill 3 — List comprehensions

The **modify-and-append** loop from the basics notebook is so common Python has a one-line form for it — shorter to read *and* faster to run:

```python
new_list = [ <transform> for item in old_list ]
new_list = [ <transform> for item in old_list if <keep?> ]   # with a filter
```

````python
# DEMO — the same result, three ways shorter
# basics notebook:  rates = []; for count in spikes: rates.append(count/2)

rates  = [count / 2 for count in spikes]          # transform every bin
active = [count for count in spikes if count > 0] # keep only firing bins

print("rates: ", rates)
print("active:", active)
````

## Skill 4 — Dict comprehensions

Same idea, but building a **dictionary**. This is the `enumerate` peaks-dict from the basics notebook, collapsed to one line:

```python
{ key: value for item in items }
```

````python
# DEMO — build dicts in one line
# basics notebook:  peaks = {}; for i, c in enumerate(spikes): peaks[i] = c

peaks    = {i: count for i, count in enumerate(spikes)}
rate_map = {i: count / 2 for i, count in enumerate(spikes) if count > 0}

print("peaks:   ", peaks)
print("rate_map:", rate_map)
````

## 🔭 Explore further — generators & iterators

Comprehensions have a lazy cousin. Swap the `[ ]` for `( )` and you get a **generator** — it produces values **one at a time, on demand**, instead of building the whole list in memory:

```python
lazy_rates = (count / 2 for count in spikes)   # nothing computed yet
next(lazy_rates)   # 1.5  — computed only now
next(lazy_rates)   # 0.0  — and now the next
```

This is exactly what a `for` loop does under the hood: it calls `iter()` on your list to get an **iterator**, then `next()` over and over until the values run out.
