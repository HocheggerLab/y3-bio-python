# 🔀 If / else — making decisions

**Time:** ~25 minutes &nbsp;•&nbsp; **Goal:** let your code **decide** what to do — bin by bin — by testing a condition inside a loop.

Same neuron recordings as the loops notebooks. A loop *visits* every bin; an `if` lets you **do something different depending on what's in it**. That combination — loop + decision — is most of data analysis.

1. A simple `if` — do something *only when* a condition holds
2. `if` / `else` — a two-way choice
3. `if` / `elif` / `else` — sort into three (or more) categories
4. Count each category — decisions + a dictionary tally

Run this first — the same two recordings as before.

## Skill 1 — A simple `if`

An `if` runs its indented body **only when** the condition is `True`:

```python
if count > 8:
    print("burst!")
```

The condition uses a **comparison operator**:

| Operator | Means | | Operator | Means |
|---|---|---|---|---|
| `>` | greater than | | `==` | **equal to** (two `=`!) |
| `<` | less than | | `!=` | not equal to |
| `>=` | greater or equal | | `<=` | less or equal |

````python
# DEMO — flag only the burst bins as we scan the recording
for count in spikes:
    if count > 8:
        print(count, "spikes — burst!")
````

## Skill 2 — `if` / `else`

`else` catches **everything the `if` didn't** — so every bin gets handled exactly once.

```python
if count > 0:
    ...        # runs when the condition is True
else:
    ...        # runs for all the rest
```

````python
# DEMO — label every bin as active or silent
for count in spikes:
    if count > 0:
        print(count, "→ active")
    else:
        print(count, "→ silent")
````

## Skill 3 — `if` / `elif` / `else`

For **more than two** outcomes, add `elif` ("else, if…") branches. Python checks them **top to bottom** and takes the **first** one that's `True`, then skips the rest — so **order matters**.

```python
if count == 0:
    label = "silent"
elif count <= 6:      # only checked if the count wasn't 0
    label = "active"
else:                 # everything left over
    label = "bursting"
```

````python
# DEMO — classify each bin by spike intensity
for count in spikes:
    if count == 0:
        label = "silent"
    elif count <= 6:
        label = "active"
    else:
        label = "bursting"
    print(count, "→", label)
````

## Skill 4 — Count each category

Now the payoff — the thing you couldn't do with loops alone. Walk the recording, **decide** which category each bin is, and **tally** it in a dictionary.

This is the accumulator pattern from the loops notebook, but with a **separate counter per category**.

````python
# DEMO — how many bins of each type?
tally = {"silent": 0, "active": 0, "bursting": 0}   # start all at zero

for count in spikes:
    if count == 0:
        tally["silent"] += 1
    elif count <= 6:
        tally["active"] += 1
    else:
        tally["bursting"] += 1

print(tally)
````
