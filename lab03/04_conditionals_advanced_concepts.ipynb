## Skill 1 — Combine conditions: `and`, `or`, `not`

One test is often not enough. Glue conditions together:

- `a and b` → **both** must be true
- `a or b`  → **either** is enough
- `not a`   → **flips** true ↔ false

```python
if count > 0 and count <= 8:   # firing, but not a burst
    ...
```

````python
# DEMO — combine conditions with and / or
for count in spikes:
    if count > 0 and count <= 8:      # BOTH must hold
        print(count, "→ firing, not a burst")
    if count == 0 or count >= 12:     # EITHER is enough
        print(count, "→ extreme bin (silent or maxed)")
````

## Skill 2 — Membership & truthiness

**`in`** tests whether a value is one of several:

```python
if count in (0, 1):   # is it 0 or 1?
    ...
```

**Truthiness** — Python already treats some values as false without you writing `== 0`:

- `0`, an empty list `[]`, an empty string `""` → **falsy**
- any non-zero number, any non-empty list → **truthy**

So `if count:` means *"if count is non-zero"*, and `if not spikes:` means *"if the recording is empty"*.

````python
# DEMO — membership with `in`, and truthiness
for count in spikes:
    if count in (0, 1):          # one of these 'basically silent' values?
        print(count, "→ quiet")

print("---")
for count in spikes:
    if count:                    # truthy = non-zero
        print(count, "is a firing bin")
    else:
        print(count, "is silent")

print("---")
recording = []
if not recording:               # empty list is falsy
    print("no data recorded!")
````

## Skill 3 — Comprehensions with `if` — filter *vs* choose

The confusing bit: `if` sits in **two different places** depending on what you want.

- **Filter** — `if` at the **end** → *keep only some* items
  ```python
  [c for c in spikes if c > 0]          # drop the silent bins
  ```
- **Choose** — `if / else` at the **front** (a one-line "ternary") → *transform every* item
  ```python
  ["active" if c > 0 else "silent" for c in spikes]
  ```

Rule of thumb: an `if` with **no `else`** filters (goes at the end); an `if/else` **chooses a value** (goes at the front).

````python
# DEMO — the two positions, side by side
firing = [c for c in spikes if c > 0]                      # FILTER  (if at the end)
labels = ["active" if c > 0 else "silent" for c in spikes] # CHOOSE  (if/else at the front)

print("filtered:", firing)
print("labels:  ", labels)

# a dict comprehension can choose too
status = {i: ("burst" if c > 8 else "normal") for i, c in enumerate(spikes)}
print("status:  ", status)
````

## Skill 4 — Finale: write your own sort 🏆

You've used `sorted()` before. Now build it yourself — proof that **loops + `if` together make an algorithm**.

We'll **rank several neurons by firing rate, highest first**, using *selection sort*:

> Repeatedly scan what's left, find the **biggest** value, move it onto the results — until nothing remains.

That's a `while` (until empty), a `for` (scan the rest), and an `if` (is this one bigger?) — everything from these two notebooks, working together.

````python
# DEMO — selection sort: rank neurons high → low, by hand
rates = [5.0, 1.5, 6.0, 0.5, 4.5]   # mean firing rate (Hz) of 5 neurons

remaining = rates.copy()   # work on a copy so the original survives
ranked = []

while remaining:                 # keep going until the pool is empty
    best = remaining[0]          # assume the first is the biggest…
    for r in remaining:
        if r > best:             # …until we find a bigger one
            best = r
    ranked.append(best)          # place the biggest onto the results
    remaining.remove(best)       # and take it out of the pool

print("original:", rates)
print("ranked:  ", ranked)
````

## 🔭 Explore further

- **`is` vs `==`.** `==` compares **values**; `is` compares **identity** (are they the *same object*?). Almost always you want `==`. The one real use of `is` is testing for `None`: `if result is None:`. Never write `count is 0` — it can silently misbehave.
- **You just reinvented `sorted()`.** `sorted(rates, reverse=True)` does your whole Skill 4 in one line — but now you know what's happening inside it.
- **Sorting by a key.** `sorted(neurons, key=...)` can rank by *any* rule — you'll meet this again in the data labs.
