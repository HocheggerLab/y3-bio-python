## Our data

```python
spikes   = [3, 0, 7, 2, 9, 0, 5, 12, 1]   # neuron A — spikes per 2-second bin
spikes_b = [4, 1, 0, 8, 6, 2, 0, 11]      # neuron B
```

## Solution — Skill 1: while loops

```python
voltage   = 0.0
threshold = 5.0
current   = 0.7    # weaker input
steps     = 0

while voltage < threshold:
    voltage += current
    steps   += 1

print(f"fired after {steps} steps (V = {voltage:.1f})")   # 8 steps (V = 5.6)
```

## Solution — Skill 2: break

```python
for i, count in enumerate(spikes_b):
    if count == 0:
        print("first silent bin:", i)      # 2
        break
```

## Solution — Skill 3: list comprehensions

```python
corrected = [count - 1 for count in spikes_b]
active_b  = [count for count in spikes_b if count > 0]

print("corrected:", corrected)
print("active_b: ", active_b)
```

## Solution — Skill 4: dict comprehensions

```python
peaks_b = {i: count for i, count in enumerate(spikes_b)}

print(peaks_b)
print("bin 3:", peaks_b[3])        # 8
```
