## Our data

```python
spikes   = [3, 0, 7, 2, 9, 0, 5, 12, 1]   # neuron A — spikes per 2-second bin
spikes_b = [4, 1, 0, 8, 6, 2, 0, 11]      # neuron B
```

## Solution — Skill 1: loop and print

```python
for count in spikes_b:
    print("bin:", count)
```

## Solution — Skill 2: loop and count

```python
total = 0
for count in spikes_b:
    total = total + count          # or: total += count
print("total spikes:", total)      # 32
print("mean per bin:", total / len(spikes_b))   # 4.0
```

## Solution — Skill 3: modify and append

```python
corrected = []
for count in spikes_b:
    corrected.append(count - 1)

print("corrected:", corrected)
print("original: ", spikes_b)      # untouched
```

## Solution — Skill 4: enumerate

```python
peaks = {}
for i, count in enumerate(spikes_b):
    peaks[i] = count

print(peaks)
print("bin 3:", peaks[3])          # 8
```
