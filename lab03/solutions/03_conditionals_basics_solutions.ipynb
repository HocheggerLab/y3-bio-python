# 🔑 Solutions — If / else — making decisions

One worked answer per exercise. There is usually more than one correct
way to write these — if yours prints the right thing, yours is right too.

Read this *after* an attempt: seeing the answer teaches far less than
getting stuck did.

## Our data

```python
spikes   = [3, 0, 7, 2, 9, 0, 5, 12, 1]   # neuron A — spikes per 2-second bin
spikes_b = [4, 1, 0, 8, 6, 2, 0, 11]      # neuron B
```

## Solution — Skill 1: a simple if

```python
for count in spikes_b:
    if count == 0:                 # == compares; a single = would assign
        print(count, "— silent bin")
```

## Solution — Skill 2: if / else

```python
for count in spikes_b:
    if count >= 5:
        print(count, "→ high")
    else:
        print(count, "→ low")
```

## Solution — Skill 3: if / elif / else

```python
for count in spikes_b:
    if count <= 1:
        label = "quiet"
    elif count <= 7:               # only reached when count > 1
        label = "medium"
    else:
        label = "loud"
    print(count, "→", label)
```

## Solution — Skill 4: count each category

```python
tally = {"silent": 0, "firing": 0}

for count in spikes_b:
    if count == 0:
        tally["silent"] += 1
    else:
        tally["firing"] += 1

print(tally)                       # {'silent': 2, 'firing': 6}
```
