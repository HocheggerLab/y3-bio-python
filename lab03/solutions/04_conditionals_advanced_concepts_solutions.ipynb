# 🔑 Solutions — If / else — advanced concepts

One worked answer per exercise. There is usually more than one correct
way to write these — if yours prints the right thing, yours is right too.

Read this *after* an attempt: seeing the answer teaches far less than
getting stuck did.

## Our data

```python
spikes   = [3, 0, 7, 2, 9, 0, 5, 12, 1]   # neuron A — spikes per 2-second bin
spikes_b = [4, 1, 0, 8, 6, 2, 0, 11]      # neuron B
```

## Solution — Skill 1: and, or, not

```python
for count in spikes_b:
    if count > 0 and count < 5:
        print(count, "→ active and below 5")

print("---")
for count in spikes_b:
    if not count == 0:             # same as: count != 0
        print(count, "→ not silent")
```

## Solution — Skill 2: membership & truthiness

```python
for count in spikes_b:
    if count in (0, 1, 2):
        print(count, "→ low")

firing = 0
for count in spikes_b:
    if count:                      # truthy = non-zero
        firing += 1
print("firing bins:", firing)      # 6
```

## Solution — Skill 3: filter vs choose

```python
bursts = [c for c in spikes_b if c > 8]                     # FILTER — if at the end
labels = ["loud" if c >= 5 else "quiet" for c in spikes_b]  # CHOOSE — if/else at the front

print("bursts:", bursts)           # [11]
print("labels:", labels)
```

## Solution — Skill 4: your own sort (low → high)

```python
neuron_rates = [2.0, 5.5, 3.0, 0.5, 4.0]

remaining = neuron_rates.copy()
ranked = []

while remaining:
    best = remaining[0]
    for r in remaining:
        if r < best:               # the one change: < finds the smallest
            best = r
    ranked.append(best)
    remaining.remove(best)

print("original:", neuron_rates)
print("ranked:  ", ranked)         # [0.5, 2.0, 3.0, 4.0, 5.5]
```
