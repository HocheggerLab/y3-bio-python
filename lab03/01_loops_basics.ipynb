## Skill 1 — Loop and print

A **loop** repeats the same action for every item in a list. The shape is always the same:

```python
for count in spikes:
    print(count)
```

- `count` is a **temporary name** — each time round, it becomes the next value in `spikes`
- the **indented** line is the *body* — it runs once per item, top to bottom
- when the list runs out, the loop stops

````python
# DEMO — visit every bin in the recording
for count in spikes:
    print(count, "spikes")
````

## Skill 2 — Loop and count

To **summarise** a recording you build up an answer as you go. This is the **accumulator** pattern:

- start a variable at `0` **before** the loop
- **add** to it inside the loop, once per item

> ℹ️ This is exactly what `sum()` does for you — but doing it by hand lets you *see* the loop working, and you can adapt it to count anything.

````python
# DEMO — total spikes across the whole recording
total = 0                      # start empty, before the loop
for count in spikes:
    total = total + count      # add this bin to the running total
print("total spikes:", total)
````

## Skill 3 — Loop, modify, and append

The real workhorse: take one list and build a **brand-new** one, transforming each value on the way. Three steps:

- start an **empty** list — `rates = []`
- inside the loop, **transform** the value
- `.append()` the result onto the new list

The original list is left untouched.

````python
# DEMO — convert each bin count to a firing rate in Hz
rates = []                     # start empty
for count in spikes:
    rates.append(count / 2)    # spikes ÷ 2 seconds = spikes per second (Hz)

print("counts:", spikes)
print("rates: ", rates)
````

## Skill 4 — `enumerate`: the position *and* the value

Often you need to know **which** bin you're on, not just its value. `enumerate()` hands you **both** at once:

```python
for i, count in enumerate(spikes):
    ...
```

- `i` is the **index** — `0, 1, 2, …` — i.e. the bin number
- `count` is the value at that position

No more keeping a separate counter by hand.

````python
# DEMO — number each bin as we go
for i, count in enumerate(spikes):
    print("bin", i, "→", count, "spikes")
````
