## 2. Keeping only the rows that match a condition

In notebook 1 you made a mask: one `True` or `False` per row. If you put a mask **inside square brackets** after `df`, pandas keeps only the rows where the mask is `True`. This is how you filter a table.

```python
df[df["Crop"] == "Raspberry"]
```

It is easiest to read from the inside out:

1. `df["Crop"]` takes the Crop column.
2. `== "Raspberry"` checks every row: is this raspberry? That gives the mask.
3. `df[ ... ]` keeps the rows where the answer was `True`.

Use `==` (two equals signs) to compare. A single `=` stores a value in a variable, which is something else.

````python
# DEMO - only the raspberry surveys
rasp = df[df["Crop"] == "Raspberry"]
print("raspberry surveys:", len(rasp))
rasp.head()
````

`len(rasp)` counts the rows that are left: 269, the same number `value_counts()` gave you in notebook 1. Notice the index on the left: the rows keep their original labels (0, 36, 37, …), so you can always tell where a row came from.

You can use any comparison: `>`, `<`, `>=`, `<=`, `==`, and `!=` (not equal).

## 3. Two conditions at once

To ask for "broad bean **and** in May", make two masks and join them with `&`, which means "and". Each condition needs its **own round brackets**:

```python
df[(df["Crop"] == "Bbean") & (df["Month"] == "May")]
```

Two things trip everyone up here, so it is worth knowing why:

- **Use `&`, not the word `and`.** Python's `and` can only compare one True/False at a time. A mask has 1,620 of them, so pandas needs `&`, which compares the two masks row by row.
- **Keep the round brackets.** Python works out `&` *before* `==`. Without brackets it tries to do `"Bbean" & df["Month"]` first, which makes no sense. The brackets make it check each condition first, then join them.

For "or" (either condition is enough), use `|` in the same way.

````python
# DEMO - raspberry surveys in August
rasp_aug = df[(df["Crop"] == "Raspberry") & (df["Month"] == "August")]
print("raspberry in August:", len(rasp_aug))
````

## 4. Several values from the same column

To keep the rows that match **any one of several** values, you could join lots of conditions with `|`. `.isin()` is shorter: give it a list, and it is `True` for every row whose value is in the list.

```python
berries = ["Raspberry", "Blackberry", "Strawberry"]
df[df["Crop"].isin(berries)]
```

````python
# DEMO - all berry surveys
berries = ["Raspberry", "Blackberry", "Strawberry"]
berry_surveys = df[df["Crop"].isin(berries)]
print("berry surveys:", len(berry_surveys))
print(berry_surveys["Crop"].value_counts())
````

## 5. Putting rows in order

`.sort_values("Column")` sorts the whole table by one column, smallest first. Add `ascending=False` to put the biggest first. Each row moves as a whole, so the other columns stay with it.

If you just want the top few, `.nlargest(3, "Column")` sorts and keeps the first three in one step.

````python
# DEMO - the three surveys with the most flowers
df.sort_values("Flowers", ascending=False).head(3)
````

These are the fruit trees in blossom from notebook 1: cherry, plum and apple, each with 2,000 flowers (estimated) in April or May.
