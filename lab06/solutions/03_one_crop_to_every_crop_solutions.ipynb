```python
# Run this first. It loads the survey you used in the lecture.
import urllib.request
from pathlib import Path

import pandas as pd

FILE = "pollinators_lecture.csv"
URL = "https://raw.githubusercontent.com/HocheggerLab/y3-bio-python/main/lab06/data/" + FILE

# On Colab the file isn't there yet, so download it once.
if not Path(FILE).exists():
    urllib.request.urlretrieve(URL, FILE)

df = pd.read_csv(FILE)
print("Loaded", FILE, df.shape)
```

```python
# DEMO - total bumblebee visits for each allotment
per_allotment = df.groupby("Allotment")["Bumblebees"].sum()
per_allotment
```

## Solution: 1. groupby: one answer per group, Write it: `per_month` and `august`

```python
per_month = df.groupby("Month")["Bumblebees"].sum()
august = per_month["August"]
per_month
```

```python
# DEMO - honeybee and bumblebee visits for every crop
visits = df.groupby("Crop")[["Honeybees", "Bumblebees"]].sum()
visits
```

## Solution: 2. Several columns at once (lecture step 2), Write it: `by_site`

```python
by_site = df.groupby("Allotment")[["Honeybees", "Bumblebees"]].sum()
by_site
# Weald (259 vs 129) and Keston (57 vs 26) have about two honeybee visits for every bumblebee visit.
```

```python
# DEMO - the bumblebee share for every crop
visits["share"] = visits["Bumblebees"] / (visits["Honeybees"] + visits["Bumblebees"])
visits.round(3)
```

## Solution: 3. A new column: the bumblebee share (lecture step 3), Write it: the `percent` column

```python
visits["percent"] = (visits["share"] * 100).round(1)
visits.head()
```

## Solution: 4. Sorting to find the top crop (lecture step 4), Write it: `ranked` and `top_crop`

```python
ranked = visits.sort_values("share", ascending=False)
top_crop = ranked.index[0]
ranked.head()
```

## Solution: 5. Can we trust it? (lecture step 5), Write it: `enough`, `trusted` and `answer`

```python
enough = visits[visits["Bumblebees"] >= 20]
trusted = enough.sort_values("share", ascending=False)
answer = trusted.index[0]
trusted
```

## Solution: 5. Can we trust it? (lecture step 5), Broken: crops with at least 20 bumblebee visits

```python
fixed = visits[visits["Bumblebees"] >= 20]   # the mask goes inside the table's brackets
fixed
```

```python
# DEMO - how many surveys of each crop?
visits["surveys"] = df.groupby("Crop").size()
visits[["Bumblebees", "surveys"]].head()
```

## Solution: 6. Comparing like with like, Write it: `bb_per_survey` and `busiest`

```python
visits["bb_per_survey"] = visits["Bumblebees"] / visits["surveys"]
busiest = visits.sort_values("bb_per_survey", ascending=False).index[0]
visits.sort_values("bb_per_survey", ascending=False).head()
```

## If you finish early, question 1

```python
visits["flowers"] = df.groupby("Crop")["Flowers"].sum()
visits["bb_per_100_flowers"] = visits["Bumblebees"] / visits["flowers"] * 100
visits.sort_values("bb_per_100_flowers", ascending=False)[["Bumblebees", "flowers", "bb_per_100_flowers"]].head()
# Courgette comes top (1.7 per 100 flowers), but from only 8 bumblebee visits to 462 flowers.
# Courgette flowers are big and few, so a handful of visits gives a high rate. Raspberry (1.1)
# is based on 311 visits and is far more solid. The flower counts are also partly estimates.
```
