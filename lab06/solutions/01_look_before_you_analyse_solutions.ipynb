```python
# Run this first. It loads the survey you used in the lecture.
import urllib.request
from pathlib import Path

import pandas as pd

FILE = "pollinators_lecture.csv"
URL = "https://raw.githubusercontent.com/HocheggerLab/y3-bio-python/main/lab06/data/" + FILE

# On Colab the file isn't there yet, so download it once.
if not Path(FILE).exists():
    urllib.request.urlretrieve(URL, FILE)

df = pd.read_csv(FILE)
print("Loaded", FILE)
```

```python
# DEMO - the size of the table, and its first rows
print(df.shape)
df.head()
```

## Solution: 1. How big is the table?, Write it: `n_rows` and `n_columns`

```python
n_rows = df.shape[0]       # the first number: rows
n_columns = df.shape[1]    # the second number: columns
print(n_rows, "rows,", n_columns, "columns")
```

```python
# DEMO - the type of every column, and whether anything is missing
df.info()
```

```python
# DEMO - a summary of the Flowers column
df["Flowers"].describe()
```

## Solution: 3. A quick summary of every number column, Write it: `median_flowers`

```python
median_flowers = df["Flowers"].median()
print(median_flowers)
```

```python
# DEMO - bumblebees: the total, the average per survey, and the most in one survey
print("total bumblebee visits:", df["Bumblebees"].sum())
print("average per survey:    ", round(df["Bumblebees"].mean(), 2))
print("most in one survey:    ", df["Bumblebees"].max())
```

## Solution: 4. One column on its own: a Series, Write it: the four totals

```python
flowers = df["Flowers"].sum()
honeybees = df["Honeybees"].sum()
hoverflies = df["Hoverflies"].sum()
most = df["Honeybees"].max()     # the biggest single value, not the total
print(flowers, honeybees, hoverflies, most)
```

## Solution: 4. One column on its own: a Series, Broken: the total hoverfly visits

```python
total = df["Hoverflies"].sum()   # capital H: column names must match exactly
print("hoverfly visits:", total)
```

```python
# DEMO - how many surveys happened in each month?
print("number of different months:", df["Month"].nunique())
df["Month"].value_counts()
```

## Solution: 5. Counting how often each value appears, Write it: `weald`, `camp_site` and `top_crop`

```python
counts = df["Allotment"].value_counts()
weald = counts["Weald"]
camp_site = counts["Camp_Site"]

# value_counts() is sorted with the most common first, so the first label is the top crop.
# Typing "Raspberry" after reading the output is fine too.
top_crop = df["Crop"].value_counts().index[0]
print(weald, camp_site, top_crop)
```

```python
# DEMO - the mask for the first five surveys, then the count over all surveys
mask = df["Honeybees"] > 5
print(mask.head())
print("surveys with more than 5 honeybees:", mask.sum())
```

## Solution: 6. Counting with a condition, Write it: `with_bumblebee` and `weald_again`

```python
with_bumblebee = (df["Bumblebees"] > 0).sum()      # or >= 1
weald_again = (df["Allotment"] == "Weald").sum()
print(with_bumblebee, weald_again)
```

## Solution: 6. Counting with a condition, Broken: surveys with more than 100 flowers

```python
big = (df["Flowers"] > 100).sum()   # brackets first, so the comparison happens before .sum()
print("surveys with more than 100 flowers:", big)
```
