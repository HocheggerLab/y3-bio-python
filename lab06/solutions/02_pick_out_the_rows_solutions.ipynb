```python
# Run this first. It loads the survey you used in the lecture.
import urllib.request
from pathlib import Path

import pandas as pd

FILE = "pollinators_lecture.csv"
URL = "https://raw.githubusercontent.com/HocheggerLab/y3-bio-python/main/lab06/data/" + FILE

# On Colab the file isn't there yet, so download it once.
if not Path(FILE).exists():
    urllib.request.urlretrieve(URL, FILE)

df = pd.read_csv(FILE)
print("Loaded", FILE, df.shape)
```

```python
# DEMO - just three columns
small = df[["Month", "Crop", "Bumblebees"]]
print(small.shape)
small.head()
```

## Solution: 1. Choosing columns, Write it: `bees`

```python
bees = df[["Crop", "Honeybees", "Bumblebees"]]   # outer brackets select, inner brackets make the list
bees.head()
```

## Solution: 1. Choosing columns, Broken: two columns at once

```python
pair = df[["Crop", "Flowers"]]   # a list of names needs its own square brackets
pair.head()
```

```python
# DEMO - only the raspberry surveys
rasp = df[df["Crop"] == "Raspberry"]
print("raspberry surveys:", len(rasp))
rasp.head()
```

## Solution: 2. Keeping only the rows that match a condition, Write it: `bbean` and `busy`

```python
bbean = df[df["Crop"] == "Bbean"]
busy = df[df["Bumblebees"] >= 3]      # "3 or more" includes 3
print(len(bbean), "broad bean surveys;", len(busy), "busy surveys")
```

```python
# DEMO - raspberry surveys in August
rasp_aug = df[(df["Crop"] == "Raspberry") & (df["Month"] == "August")]
print("raspberry in August:", len(rasp_aug))
```

## Solution: 3. Two conditions at once, Write it: `bbean_weald`

```python
bbean_weald = df[(df["Crop"] == "Bbean") & (df["Allotment"] == "Weald")]
print(len(bbean_weald))
```

## Solution: 3. Two conditions at once, Broken: broad bean in May

```python
bbean_may = df[(df["Crop"] == "Bbean") & (df["Month"] == "May")]   # & instead of and
print("broad bean in May:", len(bbean_may))
```

```python
# DEMO - all berry surveys
berries = ["Raspberry", "Blackberry", "Strawberry"]
berry_surveys = df[df["Crop"].isin(berries)]
print("berry surveys:", len(berry_surveys))
print(berry_surveys["Crop"].value_counts())
```

## Solution: 4. Several values from the same column, Write it: `n_squashes`

```python
n_squashes = len(df[df["Crop"].isin(["Squash", "Courgette"])])
# Also correct: df["Crop"].isin(["Squash", "Courgette"]).sum(), which counts the Trues in the mask
print(n_squashes)
```

```python
# DEMO - the three surveys with the most flowers
df.sort_values("Flowers", ascending=False).head(3)
```

## Solution: 5. Putting rows in order, Write it: `top3`

```python
top3 = df.nlargest(3, "Bumblebees")
# Also correct: df.sort_values("Bumblebees", ascending=False).head(3)
top3
```

## Solution: 6. Step 1 of the lecture: who visits raspberry?, Write it: `bbean_honey` and `bbean_bumble`

```python
bbean = df[df["Crop"] == "Bbean"]          # step 1: keep only broad bean
bbean_honey = bbean["Honeybees"].sum()     # step 2: add up each bee column
bbean_bumble = bbean["Bumblebees"].sum()
print(bbean_honey, "honeybee visits,", bbean_bumble, "bumblebee visits")
```
