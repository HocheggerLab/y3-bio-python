## What's in the data

In 2017 and 2018, volunteers surveyed crops on nine allotment sites in Brighton. In each **survey**, a volunteer stood at one patch of one crop for a fixed length of time and counted every insect that visited the flowers.

**One row of the table is one survey.** There are 1,620 of them.

| Column | What it means |
|---|---|
| `Month` | the month of the survey (April to September) |
| `Allotment` | which of the nine allotment sites |
| `Crop` | which crop was watched. `Rbean` is runner bean and `Bbean` is broad bean |
| `Flowers` | how many open flowers were in the patch being watched |
| `Honeybees` | how many honeybee visits were counted |
| `Bumblebees` | how many bumblebee visits were counted, all species together |
| `Hoverflies` | how many hoverfly visits were counted |
| `Terrestris` … `Hortorum` | the bumblebee visits again, split into six species (we use these in the advanced notebook) |

The question from the lecture: **if the wild bumblebees disappeared, which crop would lose the most?** Notebook 3 answers it. This notebook makes sure you know the table well enough to trust the answer.

## 3. A quick summary of every number column

`df.describe()` gives eight summary numbers for each number column. In plain English:

## 4. One column on its own: a Series

When you take one column out of a DataFrame with square brackets, you get a **Series**: a single column of values, with the row labels down the side.

```python

df["Honeybees"].sum()    # add them all up
df["Honeybees"].mean()   # the average per survey
df["Honeybees"].max()    # the most in a single survey
```

The column name goes in quotes, inside square brackets, and it must be spelled **exactly** as in the table, capital letters included.

````python
# DEMO - bumblebees: the total, the average per survey, and the most in one survey
print("total bumblebee visits:", df["Bumblebees"].sum())
print("average per survey:    ", round(df["Bumblebees"].mean(), 2))
print("most in one survey:    ", df["Bumblebees"].max())
````

An average of 0.34 bumblebees per survey means most surveys saw none at all. The total of 548 comes from a minority of busy surveys.

## 6. Counting with a condition

How many surveys saw more than five honeybees? Comparing a column with a number checks **every row at once** and gives back one `True` or `False` per row:

```python
df["Honeybees"] > 5      # True for the busy surveys, False for the rest
```

This True/False column is called a **mask**, just like the masks on the neuron in Lab 5. Python treats `True` as 1 and `False` as 0, so **adding up a mask counts the Trues**:

```python
(df["Honeybees"] > 5).sum()
```

The round brackets around the comparison matter: they make Python do the comparison first, then the sum.

````python
# DEMO - the mask for the first five surveys, then the count over all surveys
mask = df["Honeybees"] > 5
print(mask.head())
print("surveys with more than 5 honeybees:", mask.sum())
````
