That recording is just **a very long row of numbers**. Today you learn the container built for exactly that.

## Our data — one neuron, one long recording

Run this first. Every exercise below uses it.

- `dff` — the neuron's activity, one number per **frame**
- `FRAME_RATE` — 10.8 frames per second, so frame 108 is 10 seconds in

> `dff` is short for ΔF/F: *how much brighter than usual* the cell is right now. Around 0 means resting; a big positive number means a burst of activity.

````python
# Run this first. It fetches one neuron's recording and loads the trace.
import urllib.request
from pathlib import Path

import numpy as np

# Where the course data lives.
DATA_URL = "https://raw.githubusercontent.com/HocheggerLab/y3-bio-python/main/lab05/data/"

NEURON = "neuron_0588_20180828_5.csv"
if not Path(NEURON).exists():
    urllib.request.urlretrieve(DATA_URL + NEURON, NEURON)

# usecols=1 pulls out a single column -> a 1-D array.
dff = np.loadtxt(NEURON, delimiter=",", skiprows=1, usecols=1)

FRAME_RATE = 10.8   # frames per second
print("loaded", NEURON)
print(dff)
````

## Skill 1 — An array knows its own shape

A Python list will happily hold your numbers, but it can't tell you much about itself. An array can:

```python
dff.shape    # how many numbers, in what layout
dff.dtype    # what kind of numbers
```

- `.shape` is a **tuple**. `(21238,)` means 21,238 numbers in a single row — *one-dimensional*.
- `.dtype` is `float64` — these are decimals, and every single one is the same type. That uniformity is what makes arrays fast.
- `len(dff)` still works, exactly like a list.

````python
# DEMO - interrogate the array
print("shape:", dff.shape)
print("dtype:", dff.dtype)
print("how many frames:", len(dff))

# how long was the recording, in minutes?
print("minutes:", round(len(dff) / FRAME_RATE / 60, 1))
````

## Skill 2 — Indexing and slicing (you already know this)

Arrays are indexed exactly like lists — same square brackets, same zero-based counting, same negative indices:

```python
dff[0]        # the first frame
dff[-1]       # the last frame
dff[100:110]  # ten frames, starting at 100
```

Nothing new to learn here. That is deliberate: numpy borrowed the syntax you already have.

````python
# DEMO - look at pieces of the recording
print("first frame:", dff[0])
print("last frame:", dff[-1])
print("ten frames from 100:", dff[100:110])

# the first ten seconds
one_second = int(FRAME_RATE)
print("first second:", dff[:one_second])
````

## Skill 3 — Arithmetic on the whole array at once

This is where arrays stop being a fancy list. Multiply an array by a number and **every element** is multiplied:

```python
dff * 100        # every value, scaled
dff - dff.mean() # every value, shifted
```

With a list you would need a loop. With an array you write the arithmetic you mean, and numpy applies it to all 21,238 numbers.

This is called **vectorisation**, and it is the whole reason numpy exists. Notebook 4 is devoted to it.

````python
# DEMO - convert to a percentage, and centre the trace on zero
percent = dff * 100
print("as %:", percent[:5])

centred = dff - dff.mean()
print("centred:", centred[:5])

# arrays combine with each other element by element, too
print("doubled:", (dff + dff)[:5])
````

## Skill 4 — Asking the array to summarise itself 🏆

An array can describe itself in one call. These are **methods**, so they need brackets:

| method | answer |
|---|---|
| `.mean()` | the average value |
| `.max()` / `.min()` | the largest / smallest value |
| `.std()` | how spread out the values are |
| `.argmax()` | the **index** of the largest value |

`.argmax()` is the interesting one: it doesn't tell you *how* active the neuron got, it tells you **when**.

````python
# DEMO - describe the neuron in five lines
print("mean activity: ", round(dff.mean(), 4))
print("quietest:      ", round(dff.min(), 4))
print("busiest:       ", round(dff.max(), 4))
print("spread (std):  ", round(dff.std(), 4))

peak_frame = dff.argmax()
print("busiest frame: ", peak_frame)
print("...which is    ", round(peak_frame / FRAME_RATE, 1), "seconds in")
````
