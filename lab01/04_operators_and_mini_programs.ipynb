## Skill 1 — Arithmetic operators

| Operator | Meaning | Example |
|----------|---------|---------|
| `+ - * /` | usual arithmetic | `5 / 2 → 2.5` |
| `**` | power | `2 ** 10 → 1024` |
| `//` | floor division | `7 // 2 → 3` |
| `%` | remainder (modulo) | `7 % 2 → 1` |

`**` is especially useful in biology — anything that doubles, decays or grows exponentially uses it.

### 📘 Demo

```python
# DEMO — serial dilution. Each step dilutes 1:10, so after n steps the dilution is 10**n.
initial_conc_ug_ml = 1000
n_steps = 4
final_conc = initial_conc_ug_ml / (10 ** n_steps)
print(f"After {n_steps} ×10 dilutions: {final_conc} µg/mL")
```

## Skill 2 — Comparisons and boolean operators

Comparisons return a `bool`:

| Operator | Meaning |
|----------|---------|
| `==` | equal to (note: **two** equals signs) |
| `!=` | not equal |
| `<`, `<=`, `>`, `>=` | usual order comparisons |
| `and`, `or`, `not` | combine bools |

You can also chain comparisons: `10 <= x <= 100` reads exactly as it does in maths.

### 📘 Demo

```python
# DEMO — is a DNA sample concentrated enough for PCR? (target: 10–100 ng/µL)
dna_ng_ul = 47.3
ok_for_pcr = 10 <= dna_ng_ul <= 100
print(f"DNA at {dna_ng_ul} ng/µL — PCR-ready? {ok_for_pcr}")
```

## Skill 3 — Putting it together: a molarity calculator

Most useful programs are just a handful of variables, a calculation, and a formatted print statement. Here's a complete mini-program that calculates the molarity of a NaCl solution.

**Formula:** $\text{molarity (M)} = \dfrac{\text{mass (g)} / \text{MW (g/mol)}}{\text{volume (L)}}$

### 📘 Demo

```python
# DEMO — molarity of an NaCl stock
solute = "NaCl"
mass_g = 5.844
molecular_weight = 58.44     # g/mol
volume_ml = 100

volume_l = volume_ml / 1000
moles = mass_g / molecular_weight
molarity = moles / volume_l

print(f"{mass_g} g {solute} in {volume_ml} mL  →  {molarity:.3f} M")
```

**Your task:** find the missing step and fix it so the cell prints 0.100 M.

> 💡 Compare it line by line with the demo above. One line of the demo is not here.


```python
# 🐞 BUGGY — should print 0.100 M, prints 0.000 M. One line is missing.
solute = "Tris base"
mass_g = 12.114
molecular_weight = 121.14    # g/mol
volume_ml = 1000

moles = mass_g / molecular_weight
molarity = moles / volume_ml

print(f"{mass_g} g {solute} in {volume_ml} mL  →  {molarity:.3f} M")

# --- self-check (do not edit) ---
assert abs(molarity - 0.100) < 0.001, f"❌ Not yet — expected 0.100 M, got {molarity}"
print("✅ Correct — the formula wants litres, the balance sheet was in millilitres.")
```
