## Skill 4 — `bool`, and inspecting types with `type()`

A bool has only two possible values: `True` or `False` — **capitalised, and with no
quotes**. They're how Python answers yes/no questions.

If you're ever unsure what something is, call `type(value)`.

### 📘 Demo

````python
# DEMO — quality control flags for a sample
is_contaminated = False
passed_qc = True
print(is_contaminated, passed_qc)
print(type(is_contaminated))
````

### 🐞 Fix it

Two things are wrong below, and they're the two mistakes everyone makes with
booleans in their first week. Python will only show you one error at a time, so
fix it, run again, and fix the next one.

**Your task:** make the cell print `False True`, with both variables holding real
booleans — *not* text that looks like one.

> 💡 After you've fixed it, the self-check will tell you if either value is still
> secretly a string.

````python
# 🐞 BROKEN — two mistakes. Fix one, re-run, fix the next.
is_contaminated = false
passed_qc = "True"

print(is_contaminated, passed_qc)

# --- self-check (do not edit) ---
assert type(is_contaminated) is bool, "❌ is_contaminated is not a bool yet"
assert type(passed_qc) is bool, "❌ passed_qc is still a string — drop the quotes"
assert is_contaminated is False and passed_qc is True, "❌ Values should be False and True"
print("✅ Fixed — True and False are capitalised, and quotes would make them text.")
````

## 🐞 The type trap — why any of this matters

Everything above was tidy. Here is the reason types are worth a whole notebook.

Below are two cells. **One of them raises an error. The other doesn't — and that's
the dangerous one.** Before you run either, predict what each will print.

### Trap 1 — the loud failure

Run it and read the last line. Python is being helpful: it refuses to guess what
you meant by adding text to a number.

````python
# 🐞 What does this do? Predict first, then run.
reading = "3.14"      # came back from a file as text
print(reading + 1)
````

### Trap 2 — the quiet failure

You read a plate count from a spreadsheet and it arrived as **text**: `"96"`.
You have 3 plates, so you multiply to get the total number of wells.

**Expected: 288.** Run it and see what you actually get.

There is no error. The code is valid Python, it does something perfectly sensible
— just not the thing you wanted. `*` on a string means *repeat the text*.

**Your task:** fix it so `total_wells` is the number `288`.

> 💡 `int("96")` converts text to a whole number. `float("3.14")` does the same for decimals.

````python
# 🐞 BROKEN — no error, wrong answer. Should be 288.
wells_per_plate = "96"        # arrived from a spreadsheet as text
plates = 3

total_wells = wells_per_plate * plates
print("Total wells:", total_wells)

# --- self-check (do not edit) ---
assert total_wells == 288, f"❌ Not yet — expected 288, got {total_wells!r}"
print("✅ Fixed — '96' * 3 is '969696'. Valid Python, silently useless science.")
````

> **Take this one with you.** Numbers arriving as text is the single most common
> data-handling bug in biology, and you will meet it for real in Lecture 6 when
> we load a spreadsheet of pollinator counts. Today it costs you three minutes.
> In a real analysis it costs you a result you believed.
