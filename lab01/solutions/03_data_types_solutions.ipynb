## Solution — Skill 4 and the type trap

```python
is_contaminated = False          # capital F, no quotes
passed_qc = True                 # capital T, no quotes
print(is_contaminated, passed_qc)

wells_per_plate = "96"
plates = 3
total_wells = int(wells_per_plate) * plates      # convert text → int first
print("Total wells:", total_wells)

r1 = "0.48"
r2 = "0.55"
r3 = "0.56"
mean_od = round((float(r1) + float(r2) + float(r3)) / 3, 2)
print(f"Mean OD: {mean_od}")
```
