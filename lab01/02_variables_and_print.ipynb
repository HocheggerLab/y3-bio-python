### 🚨 First: how to read an error

You will break things constantly — professionals see errors dozens of times a
day. One habit separates people who get unstuck from people who don't:

> **Read the LAST line of the error first.**

Everything above it is scaffolding. The last line names the problem and usually
the exact word that caused it:

```
NameError: name 'protien_conc' is not defined
^^^^^^^^^                ^^^^^^^^^^^^^
what went wrong          which name Python couldn't find
```

| Error | Means | Usually caused by |
|-------|-------|-------------------|
| `NameError` | Python doesn't know that name | a typo, or a cell you haven't run yet |
| `SyntaxError` | The line isn't valid Python | a missing comma, bracket or quote |
| *(no error at all)* | **The most dangerous one** | code that runs and prints the wrong thing |

Watch that last row. An error is Python doing you a favour; wrong output with no
error is Python letting you walk off a cliff.

## Skill 1 — Assigning a variable

A variable is a name that points to a value. The `=` sign means *"store the value on the right under the name on the left"*.

By convention, variable names are written in `snake_case`: no capitals, words joined by underscores.

Some simple rules:

- **No spaces.** `buffer ph = 7.42` is a `SyntaxError` — use `buffer_ph`.
- **Case matters.** `Buffer_ph` and `buffer_ph` are two different names.
- **Spelling matters.** Python has no idea what you *meant*.

### 📘 Demo

````python
# DEMO — pH of a freshly prepared phosphate buffer
buffer_ph = 7.42
print(buffer_ph)
````

## Skill 3 — f-strings (formatted strings)

Comma-separated `print()` is fine, but f-strings give you full control. Two rules:

1. Put `f` immediately before the opening quote: `f"..."`.
2. Anything in `{ }` inside the string is replaced by the value of that variable.

You can also format the number: `{value:.2f}` rounds to 2 decimal places, `{count:,}` adds thousands separators.

### 📘 Demo

````python
# DEMO — formatted report for a NaCl stock solution
solute = "NaCl"
molarity = 0.15378     # mol/L
volume_ml = 500

print(f"Prepared {volume_ml} mL of {molarity:.3f} M {solute}")
````

### 🐞 Fix it

**This is the one to watch.** The cell below runs perfectly. No red text, no
error, no complaint. And the output is wrong:

```
Sample S-014 had OD600 = {od600}
```

Instead of the reading, you get the *word* `{od600}`. Python did exactly what it
was told — it just wasn't told to substitute anything.

This is the single most common f-string mistake, and because it never raises an
error it can sit in a script for weeks. Compare the line below with the demo
above, character by character, and you'll see what's missing.

### 📏 Formatting numbers inside `{ }`

So far `{ }` has just dropped a value in as it is. Add a **colon** after the name
and you can also say *how* the number should look:

```
{molarity:.3f}
 ^^^^^^^^ ^^^^
 which    how to show it
 value
```

| You write | Value | Output | Means |
|-----------|-------|--------|-------|
| `{molarity:.3f}` | `0.15378` | `0.154` | 3 decimal places (rounded) |
| `{molarity:.2f}` | `0.15378` | `0.15` | 2 decimal places |
| `{od600:.3f}` | `0.6` | `0.600` | pads with zeros to keep 3 places |
| `{cells:,}` | `1250000` | `1,250,000` | comma every three digits |
| `{volume_ul:,.1f}` | `12500.37` | `12,500.4` | both: commas **and** 1 decimal place |

Read `.3f` as "**f**ixed-point, **3** places". Anything outside the `{ }` is printed as
it stands, so units and symbols such as `mL` or `%` just go in the text.

> ⚠️ Formatting changes only what is **printed**. The variable keeps its full value:
> `molarity` is still `0.15378` afterwards.
